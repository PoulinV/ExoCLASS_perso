```python
%load_ext autoreload
import sys
sys.path.append("./DarkHistory")

import numpy as np
from tqdm import tqdm_notebook as tqdm

import darkhistory.main as main
import darkhistory.physics as phys
from plotting_utils import plot_distortion

import matplotlib.pyplot as plt
from matplotlib.pyplot import cm
%matplotlib inline
```

```python
# Dark matter model parameters
mDM = 7.2e6 # eV
lifetime = 5.9e24 # s
chan = 'elec'
process = 'decay'
```

```python
%autoreload
# Run DarkHistory for LCDM (to subtract from injection model)
result_lcdm = main.evolve(
    DM_process=process, primary=chan+'_delta',
    mDM=mDM, lifetime=lifetime*1e40,
    start_rs=3000, end_rs=4.,
    coarsen_factor=16, rtol=1e-6, nmax=10, iterations=1,
    reion_switch=True, reion_method='Puchwein', heat_switch=True,
    distort=True, fexc_switch=True, reprocess_distortion=True
)

# Run DarkHistory for injection model
result_exo = main.evolve(
    DM_process=process, primary=chan+'_delta',
    mDM=mDM, lifetime=lifetime,
    start_rs=3000, end_rs=4.,
    coarsen_factor=16, rtol=1e-6, nmax=10, iterations=1,
    reion_switch=True, reion_method='Puchwein', heat_switch=True,
    distort=True, fexc_switch=True, reprocess_distortion=True
)
```

```python
plt.figure()
ax = plt.gca()

# Plot total distortion
plot_distortion(result_exo, result_lcdm, diff=True, color='k', label='Total')

# Plot y-distortion
temp_y_lcdm = {'distortion': result_lcdm['distortion_y']}
temp_y_exo = {'distortion': result_exo['distortion_y']}
plot_distortion(temp_y_exo, temp_y_lcdm, diff=True, color='r', label='y only')

plt.show()
```
